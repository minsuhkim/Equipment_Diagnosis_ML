```python
import glob, os
```

```python
folder_path = './data/archive'

txt_paths = glob.glob(os.path.join(folder_path,"*.txt"))

txt_titles = []
for path in txt_paths:
    filename = os.path.basename(path)
    title,_ = os.path.splitext(filename)
    txt_titles.append(title)

print(txt_titles)
```

```python
sensor_dict = {}
exclude_files = ['description.txt', 'documentation.txt', 'profile.txt']

for path in txt_paths:
    file_name = os.path.basename(path)
    if file_name in exclude_files:
        continue

    df = pd.read_csv(path, sep='\s+', header = None)
    key_name = os.path.splitext(file_name)[0]
    sensor_dict[key_name] = df
```

```python
sensor_dict['CE'].head()
```

```python
profile = ['Cooler_condition', 'Valve_condition', 'Internal_pump_leakage', 'Hydraulic_accumulator', 'stable_flag']
profile_dict = {}
profile_path = './data/archive/profile.txt'
profile_df = pd.DataFrame()

with open(profile_path) as f:
    lines = f.readlines()
    df = pd.DataFrame()
    data = []
    for line in range(len(lines)):
        tmp = lines[line].split('\t')
        tmp[-1] = tmp[-1][:-1]
        data.append(tmp)
    df = pd.DataFrame(data, columns = profile)
    profile_df = df

for label in profile:
    profile_dict[label] = profile_df.loc[:,label]
```

```python
feature_list = [] # 만들어진 데이터프레임들을 담을 리스트

for sensor_name, df_sensor in sensor_dict.items():
    _mean = df_sensor.mean(axis=1)
    _std = df_sensor.std(axis=1)
    _var = df_sensor.var(axis=1)
    
    _mean.name = f"{sensor_name}_mean"
    _std.name = f"{sensor_name}_std"
    _var.name = f"{sensor_name}_var"
    
    # 3. 리스트에 추가
    feature_list.append(_mean)
    feature_list.append(_std)
    feature_list.append(_var)

# 4. 최종적으로 옆으로 쫙 이어 붙이기
X_final = pd.concat(feature_list, axis=1)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, f1_score, classification_report
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neighbors import KNeighborsClassifier 
from sklearn import svm
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate
```

```python
def perform_LR(X, y):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
    scaler = StandardScaler()
    X_Scaling = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)
    model = LogisticRegression()
    model.fit(X_Scaling, y_train)

    y_pred = model.predict(X_test)
    print(classification_report(y_test,y_pred))
    accuracy = model.score(X_test, y_test)
    print("Accuracy:", accuracy)
```

```python
for label in list(profile_dict.keys()):
    print("condition:", label)
    perform_LR(X_final, profile_dict[label])
```

```python
def perform_KNN(k):
    for condition in profile:
        X_train, X_test, y_train, y_test = train_test_split(X_final, profile_dict[condition], test_size = 0.2, random_state = 1)
        scaler = StandardScaler()
        X_Scaling = scaler.fit_transform(X_train)
        X_test = scaler.transform(X_test)
        
        model = KNeighborsClassifier(n_neighbors = k)
        model.fit(X_Scaling,y_train)
        
        y_pred = model.predict(X_test)
        print('condition:', condition)
        print(classification_report(y_test, y_pred))
        print('Accuracy:',model.score(X_test, y_test))
        print('')
```

```python
def perform_SVM(X,y, label):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
    scaler = StandardScaler()
    X_Scaling = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)
    
    model = svm.SVC(kernel = 'linear', C=1)
    model.fit(X_Scaling, y_train)
    
    y_pred = model.predict(X_test)
    print("condition:", label)
    print(classification_report(y_test,y_pred))
    print("Accuracy: ", model.score(X_test, y_test))
```

```python
for label in list(profile_dict.keys()):
    perform_SVM(X_final, profile_dict[label], label)
```

```python
def perform_RF(X,y):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
    scaler = StandardScaler()
    X_Scaling = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)
    
    model = RandomForestClassifier(n_estimators = 100, max_depth=100, random_state=0)
    model.fit(X_Scaling,y_train)
    
    y_pred = model.predict(X_test)
    print(classification_report(y_test,y_pred))
    print("Accuracy:",model.score(X_test, y_test))
```

```python
for label in list(profile_dict.keys()):
    print("condition:", label)
    perform_RF(X_final, profile_dict[label])
```

```python
def perform_RF_cv(X, y, k=5):
    # 1. 파이프라인 구축 (스케일러 + 모델)
    pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('rf', RandomForestClassifier(n_estimators=100, max_depth=100, random_state=0))
    ])
    
    # 2. K-Fold 설정 (StratifiedKFold 사용 권장)
    # 분류 문제에서는 정답(label) 비율을 유지해주는 StratifiedKFold가 좋습니다.
    cv = StratifiedKFold(n_splits=k, shuffle=True, random_state=1)
    
    # 3. 교차 검증 수행 (cross_validate)
    scoring_metrics = ['accuracy', 'f1_macro', 'precision_macro', 'recall_macro']
    scores = cross_validate(pipeline, X, y, cv=cv, scoring = scoring_metrics)
    
    # 4. 결과 출력
    print(f"=== {k}-Fold Cross Validation Results ===")
    print(f"Average Accuracy : {np.mean(scores['test_accuracy']):.4f}")
    print(f"Average Precision: {np.mean(scores['test_precision_macro']):.4f}")
    print(f"Average Recall   : {np.mean(scores['test_recall_macro']):.4f}")
    print(f"Average F1 Score : {np.mean(scores['test_f1_macro']):.4f}")
```

```python
for label in list(profile_dict.keys()):
    print("condition:", label)
    perform_RF_cv(X_final, profile_dict[label])
```
